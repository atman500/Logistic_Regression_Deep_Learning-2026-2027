```python
import numpy as np
import pandas as pd
from google.colab import files
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score, precision_score
```

```python
df=pd.read_csv('Multiple_logistic_Regression.csv',sep=';')
print(df.head())
```

```python
x=df['Pass']
y=df.drop(columns=['Pass'],axis=1)
```

```python
print(df.describe())
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn import model_selection
model_selection=LogisticRegression()
model_selection.fit=(x_train,y_test)

```

```python
from sklearn.linear_model import LogisticRegression
# 1. Initialize the model with a clear variable name
model = LogisticRegression(solver='liblinear') # Added solver for better convergence
# 2. Call the .fit() method correctly using training features and training targets
model.fit(y_train, x_train)
```

```python
# Make predictions on the test set
y_pred = model.predict(y_test)
```

```python
# Calculate and print the Confusion Matrix
conf_matrix = confusion_matrix(x_test, y_pred)
print('Confusion Matrix:')
print(conf_matrix)
```

```python
# Calculate and print Accuracy, Recall, and Precision
accuracy = accuracy_score(x_test, y_pred)
recall = recall_score(x_test, y_pred)
precision = precision_score(x_test, y_pred)

print(f'Accuracy: {accuracy:.4f}')
print(f'Recall: {recall:.4f}')
print(f'Precision: {precision:.4f}')
```
